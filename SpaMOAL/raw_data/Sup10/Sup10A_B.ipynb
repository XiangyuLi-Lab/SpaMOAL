```python
import scanpy as sc
adata = sc.read("pseudotime.h5ad")
print(adata)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = adata.obs[['no_image', 't']].copy()
filtered_df = df[df['no_image'].isin([8, 9])]
filtered_df['no_image'] = filtered_df['no_image'].astype(int)

ax = sns.boxplot(
    data=filtered_df, 
    x='no_image', 
    y='t',
    linewidth=3,  
    palette=['#4d5ba0', '#23a9ad'],
    width=0.3,
    showfliers=False,
    boxprops={'edgecolor': 'black'},
    medianprops={'color': 'black'},
    whiskerprops={'color': 'black'},
    capprops={'color': 'black'}
)

ax.spines['top'].set_linewidth(2)
ax.spines['right'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.spines['left'].set_linewidth(2)

# 加粗刻度线
ax.tick_params(axis='x', width=2)
ax.tick_params(axis='y', width=2)

ax.set_xticklabels(
    ['Cluster 8', 'Cluster 9'],
    fontsize=25
)
ax.set_ylim(0, 1.2)
ax.set_yticks([0, 0.6, 1.2])
ax.set_yticklabels(
    [0, 0.6, 1.2],
    fontsize=25
)

plt.title('Pseudotime Score',fontsize=25)
plt.xlabel('')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
