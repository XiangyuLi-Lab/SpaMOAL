```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import numpy as np

# 定义数据
labels = ['AMI', 'NMI', 'ARI', 'Homogeneity', 'V-measure', 'Mutual Information']
methods = ['STAGATE', 'SpaGCN','SpaMOAL:RNA', 'SpaMOAL']
data = {
  'STAGATE': [0.6383, 0.6408, 0.6051, 0.6393, 0.6408, 0.8548],
    'SpaGCN': [0.8099, 0.8112, 0.8333, 0.8145, 0.8112, 1.0890],
    'SpaMOAL:RNA':[0.6672, 0.6695, 0.6222, 0.6654, 0.6695, 0.8897],
    'SpaMOAL': [0.8005, 0.8019, 0.8564, 0.7997, 0.8019, 1.0692],
}

# 设置图形大小
fig, ax = plt.subplots(figsize=(32, 13))  # 加宽画布，给图例留空间
plt.rcParams['font.size'] = 18

bar_width = 0.2
index = np.arange(len(labels))

# 粉色系：第3、4个同色系深浅
colors = ['#E8A16D', '#40AE91', '#DD7499', '#D6608A']

# 绘制条形图
for i, method in enumerate(methods):
    ax.bar(index + i * bar_width, data[method], bar_width, label=method, color=colors[i])

# 图例大幅右移，远离柱子
legend = ax.legend(loc='upper right', bbox_to_anchor=(1.5, 1), borderaxespad=0.5, fontsize=65, 
          frameon=False, handlelength=1, handleheight=1, handletextpad=0.5)

# 统一字体
legend_text = legend.get_texts()[0]
legend_font = legend_text.get_fontproperties()
font_family = legend_font.get_family()

ax.set_xticks(index + bar_width * (len(methods) - 1) / 2)
ax.set_xticklabels(labels, rotation=45, ha='right', fontsize=65, family=font_family)

ax.yaxis.set_major_locator(MaxNLocator(nbins=5))
plt.yticks(fontsize=65)
ax.tick_params(axis='y', pad=20)

# 隐藏顶部和右侧边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
# bbox_inches='tight' 保存时自动裁掉空白，保证图例完整
plt.show()
```
